```python
import os, sys, copy
from pathlib import Path
ROOT=Path.cwd()
```

```python
%%time

ORIG  = ROOT / "original_files"

# parameter point (H0=67.4, omega_b=0.0224, omega_cdm=0.120 fixed)
POINT = {
    "lsum": -0.902196, "ldiff": -1.20594, "l2": 10.3442, "l3": 0.677975,
    "asum": 0.512237, "adiff": 0.342735, "log10_fesc10": -0.834616,
    "alpha_esc": -0.338567, "log10Mcrit": 9.19371,
    "n_s": 0.943396, "logA": 3.32225,
}

# Adding the old files to path
sys.path.insert(0, str(ORIG))

import mcmc_uvlf_9params_old as mcmcmod          # the pre-optimisation file

info = copy.deepcopy(mcmcmod.info)
for k in ("output", "resume", "sampler"):
    info.pop(k, None)

from cobaya.model import get_model
model = get_model(info)
res = model.logposterior(POINT, return_derived=True)
dn = list(model.parameterization.derived_params().keys())
der = res.derived if isinstance(res.derived, dict) else dict(zip(dn, res.derived))
chi = {n: -2.0 * float(v) for n, v in zip(model.likelihood.keys(), res.loglikes)}

print("log-prior      = " + str(res.logprior))
print("log-like       = " + str(res.loglike))
print("log-posterior  = " + str(res.logpost))
print("chi2 per likelihood:")
for n in model.likelihood.keys():
    print("  " + n + "  =  " + str(chi[n]))
print("tau_e = " + str(der["tau_e"]))
print("A_s   = " + str(der["A_s"]))
```
